## 1. `range()` — générer une séquence de nombres

`range()` est un outil qui permet de **générer une suite de nombres**. Python s’en sert très souvent avec les boucles.

On peut le voir comme une **liste virtuelle** de nombres :
- `range(5)` correspond à : 0, 1, 2, 3, 4
- `range(1, 5)` correspond à : 1, 2, 3, 4

On rappelle que Python commence à compter à 0.
On va d’abord regarder ce que Python affiche, puis on fera des petits exercices.


````python
# Exemple : quelques ranges
print(f"range(5) -> {list(range(5))}")          # 0,1,2,3,4
print(f"range(1, 5) -> {list(range(1, 5))}")    # 1,2,3,4
print(f"range(0, 10, 2) -> {list(range(0, 10, 2))}")  # 0,2,4,6,8
````

## 2. Boucles `for` avec `range` — répéter une action

Une **boucle `for`** permet de répéter une action **plusieurs fois**.

Exemple d’idée :
- envoyer un message pour chaque jour d’une campagne,
- afficher les numéros des emails envoyés,
- tester quelque chose pour les 5 premiers clients.

Structure générale :
```python
for nombre in range(5):
    print(nombre)
```

Dans la cellule suivante, on fait quelques boucles très simples.

````python
# Exemple : répéter une action plusieurs fois

for i in range(5):
    print(f"Envoi de l'email numéro {i}")

print(f"-")

for jour in range(1, 4):
    print(f"Jour de campagne : {jour}")
````

## 4. Combiner deux listes : clients + montants

Souvent, on a **deux listes liées** :
- une liste de clients ;
- une liste de montants d’achats ;
- à la position 0 : client 0 et montant 0 vont ensemble, etc.

On peut alors utiliser un **indice** `i` pour parcourir les deux listes en même temps.

Schéma :
```python
clients = ["Alice", "Bob"]
montants = [120, 45]

for i in range(len(clients)):
    nom = clients[i]
    montant = montants[i]
    print(nom, montant)
```

````python
# Exemple : clients + montants
clients = ["Alice", "Bob", "Chloé", "David"]
montants_achats = [120, 45, 200, 80]

for i in range(len(clients)):
    nom = clients[i]
    montant = montants_achats[i]
    print(f"Client {nom} - montant d'achat : {montant} €")
````

## 5. Conditions `if / elif / else` — prendre des décisions

Les **conditions** permettent de prendre des décisions en fonction d’une règle.

Exemples marketing :
- si le montant d’achat est supérieur à 100 € → client **premium** ;
- sinon → client **standard**.

Structure générale :
```python
if condition:
    # faire quelque chose
else:
    # faire autre chose
```

````python
# Exemple : client premium ou standard

montant_achat = 120
seuil_premium = 100

if montant_achat > seuil_premium:
    print(f"Client premium")
else:
    print(f"Client standard")
````

## 6. Mini défi — Qui relancer, qui garder ?

On combine maintenant :
- listes ;

- conditions.

### Scénario
Tu as :
- une liste de clients ;
- une liste de montants d’achats correspondants ;
- un seuil de relance (par ex. 60 €).

Règle simple :
- si un client a dépensé **moins** que le seuil → `"à relancer"` ;
- sinon → `"client intéressant"`.

### Objectif
1. Créer les listes et le seuil.
2. Parcourir les clients avec une boucle.
3. Pour chaque client, afficher un message :

   > "Client X a dépensé Y € : à relancer."  
   ou  
   > "Client X a dépensé Y € : client intéressant."

````python
# TODO Mini défi — Qui relancer, qui garder ?
# Complète `clients`, `montants_achats` et `seuil_relance`,
# puis exécute la cellule suivante (correction).

pass
````

````python
# Mini défi : relance ou non

clients = ["Alice", "Bob", "Chloé", "David"]  # tu peux modifier
montants_achats = [120, 45, 200, 80]               # tu peux modifier
seuil_relance = 60

for i in range(len(clients)):
    nom = clients[i]
    montant = montants_achats[i]

    if montant < seuil_relance:
        print(f"Client {nom} a dépensé {montant} € : à relancer.")
    else:
        print(f"Client {nom} a dépensé {montant} € : client intéressant.")
````
