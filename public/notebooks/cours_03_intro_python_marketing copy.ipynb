## 2. Les listes : créer et utiliser l’index

Une **liste** sert à stocker plusieurs valeurs, par exemple :
- des noms de campagnes
- des pays
- des devices
- des budgets

L’**index** sert à repérer une valeur dans la liste.
- l’index commence à `0`
- ensuite `1`, `2`, `3`, etc.

Ce qu’on veut savoir faire ici :
- créer une liste
- accéder à un élément via `liste[index]`
- modifier un élément : `liste[index] = nouvelle_valeur`

```python
# List basics: create a list and access items by index

campagnes = ["Google Ads", "Facebook Ads", "Email"]
budgets = [500, 300, 150]

print("campagnes[0] =", campagnes[0])
print("campagnes[1] =", campagnes[1])
print("budgets[2] =", budgets[2])
```

## 3. Accéder et modifier une liste

Maintenant qu’on sait lire un élément avec `liste[index]`, on peut aussi **modifier** une valeur.

La syntaxe ressemble à ça :
- lecture : `valeur = liste[index]`
- modification : `liste[index] = nouvelle_valeur`

Dans les exercices, on va utiliser ça pour changer un **budget** de campagne.

```python
# Modify a list item by index

campagnes = ["Google Ads", "Facebook Ads", "Email"]
budgets = [500, 300, 150]

print("Budgets avant:", budgets)

budgets[1] = 450  # update the budget for the campaign at index 1
print("Budgets après:", budgets)

print("campagnes[1] =", campagnes[1])
print("budget correspondant =", budgets[1])
```
