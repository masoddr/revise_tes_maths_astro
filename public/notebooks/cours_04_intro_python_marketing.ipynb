## 1. Introduction aux dictionnaires (`dict`)

Un dictionnaire permet de stocker des donnees avec des **cles explicites**.

Exemple mental:
- Liste: position 0, position 1, position 2
- Dictionnaire: `"nom"`, `"budget"`, `"clics"`, `"conversions"`

C est tres pratique en marketing, car les champs ont un nom metier lisible.

```python
# Example 1: create a dictionary for one campaign
campagne = {
    "nom": "Launch Produit A",
    "canal": "Google Ads",
    "budget": 1200,
    "clics": 480,
    "conversions": 36,
}

print(campagne)
print(f"Nom: {campagne['nom']}")
print(f"Budget: {campagne['budget']} euros")
```

## 2. Lire, modifier et enrichir un dictionnaire

Operations de base:
- Lire une valeur: `campagne['budget']`
- Modifier une valeur: `campagne['budget'] = 1500`
- Ajouter une cle: `campagne['objectif'] = 'Leads'`

Astuce: tu peux aussi utiliser `.get('cle', valeur_par_defaut)` pour eviter une erreur si la cle n existe pas.

```python
# Example 2: update and extend a dictionary
campagne = {
    "nom": "Promo Ete",
    "budget": 800,
    "clics": 200,
    "conversions": 14,
}

# Update one value
campagne["budget"] = 900

# Add a new key
campagne["objectif"] = "Ventes"

print(campagne)
print(f"Objectif: {campagne.get('objectif', 'Non defini')}")
```

## 3. Liste de dictionnaires: modele proche du reel

En pratique, on manipule rarement une seule campagne.
On utilise plutot une **liste de dictionnaires**:
- chaque element de la liste = une campagne
- chaque campagne = un dictionnaire de champs metier

```python
# Example 3: list of dictionaries
campagnes = [
    {"nom": "Brand FR", "budget": 1000, "clics": 400, "conversions": 24},
    {"nom": "Acquisition UK", "budget": 1500, "clics": 500, "conversions": 30},
    {"nom": "Retargeting ES", "budget": 700, "clics": 280, "conversions": 21},
]

for campagne in campagnes:
    cpc = campagne["budget"] / campagne["clics"]
    taux_conversion = campagne["conversions"] / campagne["clics"]
    print(f"{campagne['nom']} | CPC={cpc:.2f} euros | Taux conv={taux_conversion:.2%}")
```

```python
# Correction — Exercise 3
campagnes = [
    {"nom": "Campagne A", "budget": 900, "clics": 300, "conversions": 27},
    {"nom": "Campagne B", "budget": 1200, "clics": 500, "conversions": 20},
    {"nom": "Campagne C", "budget": 700, "clics": 210, "conversions": 8},
]

for campagne in campagnes:
    cpc = campagne["budget"] / campagne["clics"]
    taux_conversion = campagne["conversions"] / campagne["clics"]

    if taux_conversion >= 0.08:
        segment = "A renforcer"
    elif taux_conversion >= 0.05:
        segment = "A optimiser"
    else:
        segment = "A revoir"

    print(
        f"{campagne['nom']} | Budget={campagne['budget']} | Clics={campagne['clics']} | "
        f"Conversions={campagne['conversions']} | CPC={cpc:.2f} euros | "
        f"Taux conv={taux_conversion:.2%} | Segment={segment}"
    )
```
