## 1. Du dictionnaire au `DataFrame`

En pratique, on ne fait pas tout a la main avec des dictionnaires et des boucles : on utilise **`pandas`**, une bibliotheque qui represente les donnees sous forme de **tableau** (lignes / colonnes), comme Excel ou comme une table SQL.

L objet principal s appelle un **`DataFrame`**.

Tres important :
- une **ligne** du `DataFrame` = un **dictionnaire** (une campagne, un utilisateur, une session, ...)
- une **colonne** du `DataFrame` = une **cle** commune a toutes les lignes (`nom`, `budget`, `clics`, ...)

Donc passer d une liste de dictionnaires a un `DataFrame` est tres naturel.

````python
# Step 1: import pandas
import pandas as pd

# Step 2: build a DataFrame from a list of dictionaries
campagnes = [
    {"nom": "Brand FR",       "canal": "Google Ads", "budget": 1000, "clics": 400, "conversions": 24},
    {"nom": "Acquisition UK", "canal": "LinkedIn",   "budget": 1500, "clics": 500, "conversions": 30},
    {"nom": "Retargeting ES", "canal": "Facebook",   "budget": 700,  "clics": 280, "conversions": 21},
]

df_campagnes = pd.DataFrame(campagnes)
df_campagnes
````

Quelques observations :
- pandas a devine les **noms de colonnes** a partir des cles du dictionnaire.
- chaque **ligne** correspond a **un** dictionnaire de la liste.
- l affichage est beaucoup plus lisible qu un `print(campagnes)`.

Petite question : combien y a-t-il de lignes et combien de colonnes dans ce `DataFrame` ?

## 3. Explorer un `DataFrame` : `head`, `tail`, `shape`, `columns`, `info`

Quand on recoit un nouveau fichier, on fait **toujours** la meme chose : on regarde a quoi il ressemble.

Les outils de base :
- `df.head(n)` : les **n premieres** lignes (par defaut 5).
- `df.tail(n)` : les **n dernieres** lignes.
- `df.shape`   : un couple `(nb_lignes, nb_colonnes)`.
- `df.columns` : la liste des **noms de colonnes**.
- `df.info()`  : un resume avec les types de colonnes et le nombre de valeurs.

Parallele SQL : c est l equivalent du `SELECT * FROM sessions LIMIT 5;` ou de la commande `DESCRIBE sessions;`.

````python
# Explore the DataFrame
print("Shape (rows, columns):", df_sessions.shape)
print("Columns:", list(df_sessions.columns))

print("\nFirst 5 rows:")
print(df_sessions.head())

print("\nLast 3 rows:")
print(df_sessions.tail(3))
````

## 4. Selectionner des colonnes (parallele SQL `SELECT`)

En SQL on ecrit :
```sql
SELECT source, device, converted FROM sessions;
```

En `pandas`, on ecrit a peu pres la meme chose, mais avec des **crochets** :
- une colonne : `df['source']`
- plusieurs colonnes : `df[['source', 'device', 'converted']]`

Points importants :
- avec **une** paire de crochets et **un seul nom**, on obtient une **colonne** (= une `Series`).
- avec **deux** paires de crochets et une **liste de noms**, on obtient un **sous-`DataFrame`**.

````python
# Selection of one column
sources = df_sessions["source"]
print(type(sources))
print(sources.head())
````

````python
# Selection of several columns -> sub DataFrame
df_sub = df_sessions[["source", "device", "converted"]]
print(type(df_sub))
df_sub.head()
````

On peut faire la lecture metier suivante : 
- `df_sessions["source"]` = liste des **sources de trafic** ligne par ligne.
- `df_sessions[["source", "device", "converted"]]` = mini tableau avec les trois colonnes utiles pour analyser la **conversion par canal et par device**.
