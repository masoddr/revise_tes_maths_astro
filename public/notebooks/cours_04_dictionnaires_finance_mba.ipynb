## 1. Introduction aux dictionnaires (`dict`)

Un dictionnaire permet de stocker des données avec des **clés explicites**.

Exemple mental :
- **Liste** : position 0, position 1, position 2 → il faut se souvenir de ce qu'il y a à chaque indice.
- **Dictionnaire** : `"ticker"`, `"prix"`, `"quantite"`, `"secteur"` → chaque champ a un **nom métier** clair.

C'est très pratique en finance, car un actif a beaucoup de champs (ticker, secteur, devise, prix, quantité...) et on veut éviter de retenir l'ordre dans une liste.

````python
# Example 1: create a dictionary for one stock position
action = {
    "ticker": "AAPL",
    "secteur": "Technology",
    "devise": "USD",
    "prix": 187.50,
    "quantite": 100,
}

print(action)
print(f"Ticker : {action['ticker']}")
print(f"Secteur : {action['secteur']}")
print(f"Prix unitaire : {action['prix']} {action['devise']}")
print(f"Quantité : {action['quantite']}")

# Simple derived metric: market value of the position
valeur_marche = action["prix"] * action["quantite"]
print(f"Valeur de marché : {valeur_marche} {action['devise']}")
````

## 2. Lire, modifier et enrichir un dictionnaire

Opérations de base sur un dictionnaire :

- **Lire** une valeur : `action['prix']`
- **Modifier** une valeur : `action['prix'] = 195.30` (mise à jour suite à un mouvement de marché)
- **Ajouter une clé** : `action['pays'] = 'États-Unis'`

Ces trois opérations utilisent la même syntaxe avec des crochets `[...]`. Pour ajouter une nouvelle clé, on écrit simplement `dict['nouvelle_cle'] = valeur`.

````python
# Example 2: update and extend a dictionary (price move, new fields)
action = {
    "ticker": "MC.PA",       # LVMH on Euronext Paris
    "prix": 700.00,
    "quantite": 25,
    "devise": "EUR",
}

# Read a value
print(f"Prix actuel : {action['prix']} {action['devise']}")

# Price moved during the trading session
action["prix"] = 712.40
print(f"Prix après mise à jour : {action['prix']} {action['devise']}")

# Enrich with new fields
action["secteur"] = "Luxury"
action["actif"] = True

print(action)
````

## 3. Parcourir un dictionnaire avec `keys()`, `values()` et `items()`

Pour parcourir un dictionnaire avec une boucle `for`, on dispose de trois méthodes :

- `d.keys()` : la liste des **clés** du dictionnaire.
- `d.values()` : la liste des **valeurs** du dictionnaire.
- `d.items()` : la liste des **couples** `(clé, valeur)`.

On utilise ensuite `for cle in d.keys()`, `for valeur in d.values()` ou `for cle, valeur in d.items()` selon ce que l'on veut afficher.

C'est très utile pour faire une **fiche d'identité** d'un actif : on affiche toutes ses caractéristiques sans avoir à écrire chaque clé à la main.

````python
# Example 3: iterate with keys(), values() and items()
action = {
    "ticker": "TTE.PA",       # TotalEnergies
    "secteur": "Energy",
    "prix": 61.30,
    "quantite": 80,
    "devise": "EUR",
}

print("--- Clés du dictionnaire ---")
for cle in action.keys():
    print(cle)

print("--- Valeurs du dictionnaire ---")
for valeur in action.values():
    print(valeur)

print("--- Couples clé / valeur (items) ---")
for cle, valeur in action.items():
    print(f"{cle} : {valeur}")
````

## 4. Liste de dictionnaires : un portefeuille proche du réel

En pratique, on travaille rarement sur une seule action. On travaille sur un **portefeuille** :

- une **liste de dictionnaires**
- chaque élément de la liste = **une position** (un dictionnaire avec ses champs)
- chaque dictionnaire = un actif décrit par ses **champs métier** (`ticker`, `secteur`, `prix`, `quantite`...)

C'est la structure idéale pour boucler dessus et calculer des indicateurs (valeur, rendement, P&L).

````python
# Example 4: list of dictionaries = a small portfolio
portefeuille = [
    {"ticker": "MC.PA",  "secteur": "Luxury", "prix": 712.40, "quantite": 25},
    {"ticker": "TTE.PA", "secteur": "Energy", "prix":  61.30, "quantite": 80},
    {"ticker": "BNP.PA", "secteur": "Banks",  "prix":  68.10, "quantite": 50},
]

for position in portefeuille:
    valeur_marche = position["prix"] * position["quantite"]
    print(f"{position['ticker']} | {position['secteur']} | prix = {position['prix']} EUR | qte = {position['quantite']} | valeur = {valeur_marche} EUR")
````
