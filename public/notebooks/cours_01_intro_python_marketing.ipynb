## 1. Qu’est‑ce qu’une variable ? (explication imagée)

On va utiliser une **métaphore de “cases mémoire” dans l’ordinateur**.

- Imagine que ton ordinateur contient des **étagères** avec des **boîtes**.
- Quand tu écris `x = 5` en Python, on peut l’expliquer ainsi :
  - Python **ouvre une boîte** dans la mémoire,
  - il met la **valeur `5`** dans cette boîte,
  - il colle une **étiquette `x`** sur la boîte.
- Plus tard, quand tu écris `x + 3`, Python va chercher la boîte étiquetée `x`, lire la valeur (5), et faire le calcul.

On appelle cette opération une **assignation** :
- `x = 5` se lit : *« x reçoit la valeur 5 »*.

Dans les cellules suivantes, on va **dessiner mentalement** ce qui se passe, puis écrire du code.

```python
# Exemple 1 : première assignation
x = 5
print(f"Valeur de x : {x}")

# À VOIX HAUTE (en classe), on peut dire :
# - Python ouvre une case mémoire
# - Il met 5 dedans
# - Il colle l'étiquette x sur cette case
```

```python
# Exemple 2 : réassignation de variable
x = 5
print(f"Au début, x vaut : {x}")

x = x + 3  # on relit la case x (5), on ajoute 3, puis on remet 8 dans la case x
print(f"Après x = x + 3, x vaut : {x}")
```

## 2. Types simples : int, float, str, bool

Nous allons manipuler **quatre types de base** très utilisés en marketing :

- `int` : nombres entiers (ex : 10 visites, 350 clics)
- `float` : nombres à virgule (ex : 2.5 €, 0.03 taux de conversion)
- `str` : textes (chaînes de caractères) (ex : "Instagram", "France")
- `bool` : booléens, valeurs logiques `True` ou `False` (ex : a converti ou pas)

On va créer des variables marketing très simples de ces différents types.

```python
# Quelques exemples de types simples
nombre_visites = 10          # int : nombre entier
budget_campagne = 250.50     # float : nombre à virgule
nom_campagne = "Summer Sale"  # str : texte
conversion = True            # bool : vrai / faux

print(f"{nombre_visites} {type(nombre_visites)}")
print(f"{budget_campagne} {type(budget_campagne)}")
print(f"{nom_campagne} {type(nom_campagne)}")
print(f"{conversion} {type(conversion)}")
```

## 3. `print()` pour raconter une petite histoire marketing

`print()` sert à **afficher** des informations à l’écran.

Deux façons courantes de mélanger texte + variables :
- `print("Le client", nom_client, "a dépensé", montant, "euros")`
- `print(f"Le client {nom_client} a dépensé {montant} euros")`

Les deux font la même chose, mais la deuxième (avec `f"..."`) est souvent plus lisible.

```python
# Exemple : phrase marketing avec virgules et f-string
nom_client = "Alice"
visites_30_jours = 5
montant_achat = 120

# Version avec virgules
print(f"Le client {nom_client} a effectué {visites_30_jours} visites et a dépensé {montant_achat} euros.")

# Version avec f-string
print(f"Le client {nom_client} a effectué {visites_30_jours} visites et a dépensé {montant_achat} euros.")
```

## 4. Calculs marketing très simples (CPC, taux de conversion)

On va utiliser Python comme une **calculatrice lisible**.

Exemples de KPI :
- **CPC (Coût Par Clic)** : `budget / nombre_de_clics`
- **Taux de conversion** : `nombre_de_conversions / nombre_de_clics`

On reste volontairement sur **des cas très simples** pour bien voir le lien entre les formules et le code.

```python
# Exemple : calcul d'un CPC et d'un taux de conversion
budget = 500       # euros
clics = 250        # nombre de clics
conversions = 25   # nombre de conversions

cpc = budget / clics
taux_conversion = conversions / clics

print(f"Budget : {budget} euros")
print(f"Clics : {clics}")
print(f"Conversions : {conversions}")
print(f"CPC : {cpc} euros")
print(f"Taux de conversion : {taux_conversion}")
```

## 5. Les listes (sans boucles pour l’instant)

Une **liste** permet de stocker plusieurs valeurs dans **une seule variable**.

Exemples de listes :
- liste de noms de clients ;
- liste de montants d’achats ;
- liste de pays ou de canaux d’acquisition.

Important :
- En Python, le **premier élément** a l’indice **0** ;
- Si une liste contient 4 éléments, le dernier a l’indice **3**.

```python
# Exemple : liste de clients et montants
clients = ["Alice", "Bob", "Chloé", "David"]
montants_achats = [120, 45, 200, 80]

print(f"Liste des clients : {clients}")
print(f"Montants d'achats : {montants_achats}")

# Accéder à certains éléments
print(f"Premier client : {clients[0]}")
print(f"Dernier client : {clients[3]}")  # 3 car la liste a 4 éléments (0,1,2,3)
```

## 6. Longueur d’une liste avec `len()`

`len()` permet de connaître le **nombre d’éléments** dans une liste.

En marketing, cela peut signifier :
- nombre total de clients dans une campagne ;
- nombre de canaux d’acquisition ;
- nombre d’achats enregistrés.

```python
# Exemple : utiliser len() sur une liste de canaux
canaux = ["Instagram", "Google Ads", "Email", "SEO"]

nombre_canaux = len(canaux)
print(f"Canaux : {canaux}")
print(f"Nombre de canaux : {nombre_canaux}")
```
