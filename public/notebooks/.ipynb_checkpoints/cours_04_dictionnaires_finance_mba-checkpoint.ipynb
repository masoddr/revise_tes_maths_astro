## 0. Rappel express des notions déjà vues

Avant de commencer les dictionnaires, vérifie que tu es à l'aise avec :

- Types (`int`, `float`, `str`, `bool`)
- Listes et indices (`portefeuille[0]`, `len(portefeuille)`)
- Boucles `for` et `range`
- Conditions `if / elif / else`

Dans la cellule suivante, on fait un mini récap en une seule exécution, sur un cas finance.

````python
# Quick recap: variables, list, loop, condition (finance use-case)
prix_achat = 150.00      # purchase price per share, in EUR
prix_actuel = 162.50     # current market price per share, in EUR
quantite = 100           # number of shares held

tickers = ["AAPL", "MC.PA", "TTE.PA"]

rendement = (prix_actuel - prix_achat) / prix_achat
pnl = (prix_actuel - prix_achat) * quantite

print(f"Rendement : {rendement:.2%}")
print(f"P&L latent : {pnl:.2f} EUR")

for ticker in tickers:
    print(f"Actif suivi : {ticker}")

if rendement >= 0.10:
    print("Performance : très bonne")
elif rendement >= 0.00:
    print("Performance : positive")
else:
    print("Performance : négative")
````

## 1. Introduction aux dictionnaires (`dict`)

Un dictionnaire permet de stocker des données avec des **clés explicites**.

Exemple mental :
- **Liste** : position 0, position 1, position 2 → il faut se souvenir de ce qu'il y a à chaque indice.
- **Dictionnaire** : `"ticker"`, `"prix"`, `"quantite"`, `"secteur"` → chaque champ a un **nom métier** clair.

C'est très pratique en finance, car un actif a beaucoup de champs (ticker, secteur, devise, prix, quantité...) et on veut éviter de retenir l'ordre dans une liste.

````python
# Example 1: create a dictionary for one stock position
action = {
    "ticker": "AAPL",
    "secteur": "Technology",
    "devise": "USD",
    "prix": 187.50,
    "quantite": 100,
}

print(action)
print(f"Ticker : {action['ticker']}")
print(f"Secteur : {action['secteur']}")
print(f"Prix unitaire : {action['prix']} {action['devise']}")
print(f"Quantité : {action['quantite']}")

# Simple derived metric: market value of the position
valeur_marche = action["prix"] * action["quantite"]
print(f"Valeur de marché : {valeur_marche:.2f} {action['devise']}")
````

## 2. Lire, modifier et enrichir un dictionnaire

Opérations de base sur un dictionnaire :

- **Lire** une valeur : `action['prix']`
- **Modifier** une valeur : `action['prix'] = 195.30` (mise à jour suite à un mouvement de marché)
- **Ajouter une clé** : `action['pays'] = 'États-Unis'`

Astuce : tu peux aussi utiliser `.get('cle', valeur_par_defaut)` pour éviter une erreur si la clé n'existe pas.
Très utile, par exemple, pour gérer un champ optionnel comme `"dividende"`.

````python
# Example 2: update and extend a dictionary (price move, new fields)
action = {
    "ticker": "MC.PA",       # LVMH on Euronext Paris
    "prix": 700.00,
    "quantite": 25,
    "devise": "EUR",
}

# Price moved during the trading session
action["prix"] = 712.40

# Enrich with new fields
action["secteur"] = "Luxury"
action["actif"] = True  # True if position is still open

print(action)
print(f"Secteur : {action.get('secteur', 'Non défini')}")
print(f"Dividende annuel : {action.get('dividende', 'Non renseigné')}")
````

## 3. Liste de dictionnaires : un portefeuille proche du réel

En pratique, on travaille rarement sur une seule action. On travaille sur un **portefeuille** :

- une **liste de dictionnaires**
- chaque élément de la liste = **une position** (un dictionnaire avec ses champs)
- chaque dictionnaire = un actif décrit par ses **champs métier** (`ticker`, `secteur`, `prix`, `quantite`...)

C'est la structure idéale pour boucler dessus et calculer des indicateurs (valeur, rendement, P&L, allocation).

````python
# Example 3: list of dictionaries = a small portfolio
portefeuille = [
    {"ticker": "MC.PA",  "secteur": "Luxury", "prix": 712.40, "quantite": 25},
    {"ticker": "TTE.PA", "secteur": "Energy", "prix":  61.30, "quantite": 80},
    {"ticker": "BNP.PA", "secteur": "Banks",  "prix":  68.10, "quantite": 50},
]

for position in portefeuille:
    valeur_marche = position["prix"] * position["quantite"]
    print(
        f"{position['ticker']:<7s} | {position['secteur']:<7s} | "
        f"prix = {position['prix']:>7.2f} EUR | qte = {position['quantite']:>4d} | "
        f"valeur = {valeur_marche:>10.2f} EUR"
    )
````

## 5. Mini défi de fin de séance

Tu disposes d'un portefeuille de **5 positions** (à créer toi-même, comme liste de dictionnaires).
Chaque position doit avoir les clés : `ticker`, `prix_achat`, `prix_actuel`, `quantite`.

Trouve et affiche la **meilleure position** selon le score suivant :

```
score = (prix_actuel - prix_achat) / prix_achat   # = rendement de la position
```

**Contraintes** :
- gérer le cas où `prix_achat = 0` (éviter la division par zéro) ;
- afficher le **score** (en %) de chaque position dans une boucle ;
- afficher à la fin le **Top 1** : ticker + score + P&L latent en euros.

````python
# TODO — Mini challenge
pass
````

````python
# Correction — Mini challenge: best position by return
portefeuille = [
    {"ticker": "MC.PA", "prix_achat": 680.0, "prix_actuel": 712.4, "quantite": 25},
    {"ticker": "TTE.PA","prix_achat":  58.0, "prix_actuel":  61.3, "quantite": 80},
    {"ticker": "BNP.PA","prix_achat":  75.0, "prix_actuel":  68.1, "quantite": 50},
    {"ticker": "SAN.PA","prix_achat":  95.0, "prix_actuel":  92.2, "quantite": 40},
    {"ticker": "AIR.PA","prix_achat": 120.0, "prix_actuel": 145.5, "quantite": 30},
]

# Track the best position while iterating
meilleur_ticker = None
meilleur_score = None
meilleur_pnl = 0.0

for position in portefeuille:
    if position["prix_achat"] == 0:
        print(f"{position['ticker']:<7s} | prix_achat = 0, score non calculable")
        continue

    score = (position["prix_actuel"] - position["prix_achat"]) / position["prix_achat"]
    pnl = (position["prix_actuel"] - position["prix_achat"]) * position["quantite"]

    print(f"{position['ticker']:<7s} | score = {score:>7.2%} | P&L = {pnl:>9.2f} EUR")

    # Update the best position if this one is better (or first computed)
    if meilleur_score is None or score > meilleur_score:
        meilleur_ticker = position["ticker"]
        meilleur_score = score
        meilleur_pnl = pnl

print("-")
print(
    f"Top 1 : {meilleur_ticker} | score = {meilleur_score:.2%} | "
    f"P&L = {meilleur_pnl:.2f} EUR"
)
````
