## 1. Révision guidée : boucles + conditions marketing

Avant de découvrir `pandas`, on renforce encore **le schéma de base** :

> liste → boucle → condition → message business

C’est exactement ce qu’on fera plus tard avec des `DataFrame`.

On repart sur un mini scénario de relance.

````python
# Exemple : qui relancer, qui garder ?

clients = ["Alice", "Bob", "Chloé", "David"]
visites_30_jours = [1, 5, 0, 4]
seuil_relance = 2

for i in range(len(clients)):
    nom = clients[i]
    visites = visites_30_jours[i]

    if visites < seuil_relance:
        print(f"Client {nom} a {visites} visite(s) : à relancer.")
    else:
        print(f"Client {nom} a {visites} visite(s) : client engagé.")
````

## 2. De la liste au tableau : pourquoi `pandas` ?

Jusqu’ici, nous avons :
- des **listes** (`clients`, `montants`, `visites`...) ;
- des **boucles** pour parcourir ces listes ;
- des **conditions** pour décider quoi faire.

C’est très bien pour des **petites données**.

Mais bientôt, nous allons travailler sur des **fichiers CSV ShopNow** avec :
- des centaines ou des milliers de lignes ;
- beaucoup de colonnes (pays, device, source, revenu, etc.).

Manipuler tout ça avec des listes devient vite **compliqué**.

`pandas` apporte une structure centrale : le **`DataFrame`**.
- On peut le voir comme un **tableau Excel dans Python**.
- Chaque colonne a un nom (`source`, `device`, `revenue`, ...).
- Chaque ligne correspond à une observation (un client, une session, une commande, ...).

Dans ce Cours 3, on ne charge pas encore de CSV : on **fabrique de tout petits tableaux à la main** pour comprendre l’idée.

````python
# Importer pandas (bibliothèque pour les tableaux de données)
import pandas as pd

print("pandas est importé avec l'alias pd.")
````

## 3. Créer un petit `DataFrame` à la main

On va créer un **mini tableau** de sessions marketing, directement dans le code.

Idée :
- une liste pour la colonne `source` ;
- une liste pour la colonne `device` ;
- une liste pour la colonne `pages_viewed`.

Puis on met tout ça dans `pd.DataFrame({...})`.

C’est comme si on prenait plusieurs colonnes Excel et qu’on les collait ensemble dans un tableau Python.

````python
# Exemple : mini tableau de sessions

sources = ["Instagram", "Google Ads", "Email", "Instagram"]
devices = ["mobile", "desktop", "mobile", "desktop"]
pages_viewed = [5, 3, 8, 2]

sessions_df = pd.DataFrame({
    "source": sources,
    "device": devices,
    "pages_viewed": pages_viewed,
})

sessions_df
````

## 4. Découvrir quelques méthodes de base (`head`, `info`, `len`)

Même sur un tout petit `DataFrame`, on peut commencer à utiliser des **méthodes de base** :

- `len(df)` : nombre de lignes (comme sur une liste) ;
- `df.head()` : les premières lignes ;
- `df.info()` : un résumé du tableau (types de colonnes, nombre de valeurs non nulles).

On va tester ça sur notre `sessions_df`.

````python
# Quelques méthodes de base sur sessions_df

print("Nombre de lignes :", len(sessions_df))
print("-")
print("Premières lignes :")
print(sessions_df.head())
print("-")
print("Info sur le DataFrame :")
print(sessions_df.info())
````

## 5. Lien explicite avec les listes et les boucles

Ce que fait `pandas` peut souvent être vu comme une **version plus pratique** de :

- plusieurs listes placées côte à côte ;
- des boucles qui parcourent ligne par ligne.

Pour l’instant, retiens :
- une **ligne** d’un `DataFrame` ≈ un "client" ou une "session" ;
- une **colonne** ≈ une variable (source, device, budget, clics, ...).

Dans les prochains cours, nous utiliserons `pandas` pour faire :
- des **filtres** (comme `WHERE` en SQL) ;
- des **agrégations** (`groupby`) pour calculer des KPIs par source, device, pays, etc.

Mais avant ça, il faut déjà être à l’aise avec :
- les petites listes,
- les boucles `for`,
- les conditions `if`,
- et ces premiers `DataFrame` jouets.

## 6. Mini défi de fin de séance — Résumé marketing simple

### Scénario
Tu travailles pour ShopNow et tu veux faire un **mini résumé** de quelques campagnes.

Tu as :
- une petite table `campagnes_df` avec `campagne`, `budget`, `clics` ;
- tu veux afficher, pour chaque campagne, un petit texte marketing.

### Objectif
1. Pour chaque ligne de `campagnes_df`, construire une phrase du type :

   > "La campagne X a un budget de Y euros et a reçu Z clics."

2. Bonus : calculer un CPC approximatif `budget / clics` (si `clics` > 0) et l’afficher dans la phrase.

Dans la cellule suivante, on te donne une structure de code à compléter.

````python
# Mini défi : résumer les campagnes

# On part du DataFrame campagnes_df que tu as rempli plus haut

for index in range(len(campagnes_df)):
    campagne = campagnes_df.loc[index, "campagne"]
    budget = campagnes_df.loc[index, "budget"]
    clics = campagnes_df.loc[index, "clics"]

    if clics > 0:
        cpc = budget / clics
        print(f"La campagne {campagne} a un budget de {budget} € et a reçu {clics} clics (CPC ≈ {cpc} €).")
    else:
        print(f"La campagne {campagne} a un budget de {budget} € et 0 clic : à investiguer !")
````
